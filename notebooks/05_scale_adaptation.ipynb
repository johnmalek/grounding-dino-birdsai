# Reproduction step 5: scale adaptation

This stage compares native inference, global 1024/1706 resizing, and 384-pixel
tiled inference on validation. Submit it after B2 epoch 1 is available.


```python
from pathlib import Path
import os

REPO = Path(os.environ.get("BIRDSAI_REPO", Path.cwd())).resolve()
OUTPUT_ROOT = Path(
    os.environ.get("BIRDSAI_OUTPUT_ROOT", REPO / "outputs")
).resolve()

print(f"cd {REPO}")
print("sbatch slurm/experiment_b_scale_validation.slurm")
```

```python
import json

root = OUTPUT_ROOT / "experiment_b" / "scale_validation"

for name in (
    "B2_epoch1_native",
    "B2_epoch1_1024",
    "B2_epoch1_tiled_384",
):
    metrics_path = root / name / "metrics.json"
    if metrics_path.is_file():
        report = json.loads(metrics_path.read_text())
        print(
            name,
            "mAP=",
            report["coco"]["overall"]["AP_50_95"],
            "F1=",
            report["threshold_metrics"]["overall"]["f1"],
        )
    else:
        print("pending:", name)
```

Freeze the best validation strategy in
