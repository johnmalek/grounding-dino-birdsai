# Reproduction step 3: Experiment A zero-shot baseline

The full validation and TestReal runs use Slurm. Submit the jobs from the
repository root after preparing the COCO splits.


```python
from pathlib import Path
import os

REPO = Path(os.environ.get("BIRDSAI_REPO", Path.cwd())).resolve()
OUTPUT_ROOT = Path(
    os.environ.get("BIRDSAI_OUTPUT_ROOT", REPO / "outputs")
).resolve()

print(f"cd {REPO}")
print("sbatch slurm/experiment_a_validation.slurm")
print("sbatch slurm/experiment_a_test.slurm")
```

```python
import json

validation = OUTPUT_ROOT / "experiment_a" / "validation"

for prompt in (
    "person_animal",
    "human_animal",
    "person_wildlife",
    "person_species",
):
    metrics = validation / prompt / "metrics.json"
    if metrics.is_file():
        report = json.loads(metrics.read_text())
        print(prompt, report["coco"]["overall"]["AP_50_95"])
    else:
        print("pending:", prompt)
```

The selected prompt is recorded in
`configs/experiment_a_selected.json`. Do not tune the prompt or threshold on
TestReal.
