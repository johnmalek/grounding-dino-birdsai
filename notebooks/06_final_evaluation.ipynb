```python
from pathlib import Path
import os
import subprocess

REPO = Path(os.environ.get("BIRDSAI_REPO", Path.cwd())).resolve()
OUTPUT_ROOT = Path(
    os.environ.get("BIRDSAI_OUTPUT_ROOT", REPO / "outputs")
).resolve()

print(f"cd {REPO}")
print("sbatch slurm/experiment_b_test.slurm")
```

```python
import json

result_dir = (
    OUTPUT_ROOT
    / "experiment_b"
    / "test"
    / "B2_epoch1_tiled_384"
)

metadata = json.loads((result_dir / "metadata.json").read_text())
metrics = json.loads((result_dir / "metrics.json").read_text())

print("Complete:", metadata["is_complete"])
print(
    "Images:",
    metadata["completed_images"],
    "/",
    metadata["requested_images"],
)
print("mAP:", metrics["coco"]["overall"]["AP_50_95"])
print("AP50:", metrics["coco"]["overall"]["AP_50"])
print("F1:", metrics["threshold_metrics"]["overall"]["f1"])

assert metadata["is_complete"]
assert (result_dir / "predictions.json").is_file()
```

```python
subprocess.run(
    [
        "python",
        str(REPO / "scripts" / "summarize_results.py"),
        "--output-root",
        str(OUTPUT_ROOT),
    ],
    check=True,
)
```
