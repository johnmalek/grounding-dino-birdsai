# Reproduction step 2: prepare BIRDSAI

This notebook runs the same data-preparation script used by
`slurm/prepare_birdsai.slurm`. Set `BIRDSAI_DATASET_ROOT` to the directory
containing `TrainReal/` and `TestReal/`.


```python
from pathlib import Path
import os
import subprocess

REPO = Path(os.environ.get("BIRDSAI_REPO", Path.cwd())).resolve()
DATASET_ROOT = Path(
    os.environ.get("BIRDSAI_DATASET_ROOT", REPO / "data" / "birdsai")
).resolve()
OUTPUT_ROOT = Path(
    os.environ.get("BIRDSAI_OUTPUT_ROOT", REPO / "outputs")
).resolve()

assert (DATASET_ROOT / "TrainReal").is_dir()
assert (DATASET_ROOT / "TestReal").is_dir()

print("Repository:", REPO)
print("Dataset:", DATASET_ROOT)
print("Outputs:", OUTPUT_ROOT)
```

```python
subprocess.run(
    [
        "bash",
        str(REPO / "scripts" / "run_data_pipeline.sh"),
        str(DATASET_ROOT),
        str(OUTPUT_ROOT),
    ],
    check=True,
)
```

```python
import json

for split in ("train", "validation", "test"):
    path = OUTPUT_ROOT / "coco" / f"{split}.json"
    payload = json.loads(path.read_text())
    print(
        split,
        "images=", len(payload["images"]),
        "annotations=", len(payload["annotations"]),
    )

assert (OUTPUT_ROOT / "splits" / "split_report.json").is_file()
print("Data preparation outputs verified.")
```
