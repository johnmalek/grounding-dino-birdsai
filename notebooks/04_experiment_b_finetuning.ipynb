# Reproduction step 4: Experiment B fine-tuning

This runbook covers the B1/B2/B3 training array, epoch-1 validation, and the
three-epoch B2 continuation.


```python
from pathlib import Path
import os

REPO = Path(os.environ.get("BIRDSAI_REPO", Path.cwd())).resolve()
OUTPUT_ROOT = Path(
    os.environ.get("BIRDSAI_OUTPUT_ROOT", REPO / "outputs")
).resolve()

print(f"cd {REPO}")
print("sbatch slurm/experiment_b_train.slurm")
print("sbatch slurm/experiment_b_validation.slurm")
print("sbatch slurm/experiment_b_continue_b2.slurm")
```

```python
import json

for configuration in ("B1", "B2", "B3"):
    metadata_path = (
        OUTPUT_ROOT
        / "experiment_b"
        / "full"
        / configuration
        / "metadata.json"
    )
    if metadata_path.is_file():
        metadata = json.loads(metadata_path.read_text())
        print(
            configuration,
            metadata["status"],
            metadata["training_state"],
        )
    else:
        print("pending:", configuration)
```

Keep the tested physical batch size at 2 with gradient
accumulation 4 on the 40 GiB MIG GPU. A larger physical batch caused an
out-of-memory failure on a worst-case image.
